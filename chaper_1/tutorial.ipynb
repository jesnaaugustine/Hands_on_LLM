```python
model_name="microsoft/Phi-3-mini-4k-instruct"
tokenizer=AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(model_name,torch_dtype='auto',trust_remote_code=True,device_map ='mps')


tokenizer = AutoTokenizer.from_pretrained(model_name)


```

```python
tokens = tokenizer('Hellow world',return_tensors='pt').to('mps')
tokens
```
